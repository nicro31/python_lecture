# Simple Variables

## Introduction

As seen in the previous examples, a Python program if fundamentally a **calculator**. You can use it to perform all basic mathematical operations:

`````python
545445 + 545445
344864 * 465445
153454 - 534537
564560 / 145635
`````

As you can see, the Jupyter Notebook shows only the result of the last calculation. To see the results of all calculations, you would need to add some `print()` functions everywhere:

`````python
print(545445 + 545445)
print(344864 * 465445)
print(153454 - 534537)
print(564560 / 145635)
`````

It works just fine, but it is not very convenient! Indeed, you probably don't need to see all the results of your operations in a program. And more critically, how would you do to see the results of your calculation in another cell of the notebook (or part of your program in general), or to re-use the results to perform other calculations? Well, one option is to perform the calculations again, and each time we need them, but as you can imagine this is definitely not an efficient solution! 

Instead, we are going to store the results of our calculations in **variables**. Variables are pieces of the computer memory where we are going to store all the data we need to manipulate for our program to work. These data can be numbers of course, but also text, or even molecular structures! For the latter though, we need to resort to the more advanced concept of **objects** in Python, something that we will see later in this lecture. 

For now, let's discover the basic variables we can use in Python!

## Float

**Float** (or floating-point numbers) are used to manipulate real numbers (in its mathematical sense): basically any number you would write with a comma or dot (depending if you are a french or english speaker 😉) to indicate the decimal part of a number. Note (sorry frenchies!) that in Python, these numbers are written with a dot only, a comma would give you an error.

It is straightforward to define (**declare** to be precise) such a float variable in python:

`````python
a = 2.5
`````

As you can see, you only need to indicate a **name** for the variable, `a` in that case, and give a **value** to the variable, `534153.545634` here, using the `=` sign (`=` is called the assignment operator, note that the spaces are not mandatory but give a clearer code overall). Python will automatically recognize that `534153.545634` is a real number, and therefore will create a variable `a` of **type** float. You can verify this using the `type()` function in Python:

`````python
type(a)
`````

Now that the variable `a` has been defined, it is stored in the computer memory and we can use it everywhere we need it in our program:

`````python
print(a)
`````

`````python
3*a
`````

It's value can be changed at any time, using again the `=` sign:

`````python
a = 7.0
`````

`````python
a
`````

If you like, the variable can be directly defined as the result of an operation:

`````python
a = 3.5 * 2.5
`````

`````python
a
`````

New variables may also be created based on the `a` variable: they will have the same type and value, only a different name.

`````python
b = a
`````

`````python
type(b)
`````

`````python
b
`````

Also keep in mind that the two variables `a` and `b` defined like this are fully independant afterwards: if you change the value of `a`, variable `b` remains unnaffected, and vice-versa:

`````python
a = 3.2
b = a

print(a)
print(b)
`````

`````python
a = 5.0

print(a)
print(b)
`````

At the end of the day, you can see the computer memory as a huge chest of drawers, where each drawer can be used as variable. When you declare a **variable**, the computer will select a drawer with the right size depending on the variable **type**, attach a label with the variable **name** on the drawer handle to know what is inside, and put the actual **value** of the variable within the drawer. Whenever you need to access the variable, the computer will just open the drawer for you!

## Int

Integers are another kind of numbers that we often need to manipulate in Python, because they can be used to "count" things. They can be stored in variables with the type **int**, and are declared the same way as float. The only difference is that of course there is no dot to represent the decimal part:

`````python
a = 2
`````

`````python
a
`````

`````python
type(a)
`````

Note that just adding a dot after the integer declares a float, not an int:

`````python
b = 2.
`````

`````python
b
`````

`````python
type(b)
`````

It could be tempting to represent all numbers in a Python program using float, because they seem more flexible. If you need to manipulate continuous physicochemical properties, say for instance solvent density or viscosity, of course you will use float variables. But if you want to "count" things, say number of carbon atoms in a molecule for instance, you must use int or you may run into problem. Indeed, without entering the technical details, keep in mind that  **int are exact representations while float are approximations**! Here is an example using the `**` operator, which stands for exponentiation (`a ** b` means `a` raised to the power of `b`):

`````python
a = 10**16
print(a + 1)
`````

`````python
b = 10.0**16
print("{:.1f}".format(b + 1))
`````

As you can see the result is wrong using float due to precision error! Note that `"{:.1f}".format()` is a syntax that tells Python to explicity print the number with one digit after the dot,but we will come back to print formatting later.

`//` and `%` are other mathematical operators involving integers in Python that you should know:

`````python
print( 7 // 2 )
print( 7 % 2 )
`````

They are respectively the integer division of 7 by 2, and 7 modulo 2. Indeed, 7 = 2 x 3 + 1 

Finally, since both int and float represent numbers, one can convert an int into a float with the simple `float()` function:

`````python
a = 2
print(type(a))
print(a)

a = float(a)
print(type(a))
print(a)
`````

The other way around is also possible, but they are several options to "convert" a float into an int. The `int()` function will return the float number without its decimal part: 

`````python
a = 2.7
b = -0.3

c = int(a)
print(c)

d = int(b)
print(d)
`````

Or we can decide if we want to round down or round up the float, using the `floor()` and `ceil()` functions. These functions are located in the Python library called `math`, that we need to `import` before using (a section dedicated to Python libraries will come later in the lecture):

`````python
import math

a = 2.7

b = math.floor(a)
print(b)

b = math.ceil(a)
print(b)
`````

## String

Excellent, you are now an expert to manipulate numbers in Python! 👍

But this might not be enough to represent everything you need in your program, because very often we need to name things (e.g. if your program deals with molecules you would like to use their names) or interact with the program user (e.g. ask questions and inputs). Therefore we need to manipulate some text, which is achieved using the **string** type. They are declared using simple quotes or double quotes:

`````python
a = 'methanol'
b = "ethanol"

print(type(a))
print(type(b))
print(a)
print(b)
`````

The most common operation with strings is the concatenation, achieved with the `+` operator:

`````python
c = a + " and " + b + " are alcohols"
print(c)
`````

We are going to see many other possible operations in the next lecture session on containers. So far, just remember that strings can be edited very precisely and we can do almost everything with them. Here is small example of text editing: 

`````python
print(c.upper())
print(c.capitalize())
`````

Some special characters, starting with the **escape character** `\`, are recognize as some "special editing". The most common is the "new line" character `\n`:

`````python
print("methanol\nethanol")
`````

Just like float and int can be converted one into another, string can also be converted into numbers and numbers can be converted in string using the `str()` function:

`````python
a = 1.2
b = 3

c = str(a)
print(type(c))
print(c)

d = str(b)
print(type(d))
print(d)
`````

## Bool

Boolean are the last kind of primitive Python variables, encoded with the **bool** type. They are very simple as they can only represent two values, either `True` or `False`:

`````python
a = True
print(type(a))
print(a)

b = False
print(type(b))
print(b)
`````

They can be converted to numbers and strings, and vice-versa with the `bool()` function, with the following rule: empty strings and 0 are equivalent to `False`, any other values are equivalent to `True`. The string conversion of a bool just gives you the boolean as a text: 

`````python
print( bool("") )
print( bool("ethanol") ) 

print( bool(0.0) )
print( bool(-2.5) )
print( bool(1) )

print( float(True) )
print( int(False) )
print( str(True) )
print( str(False) )
`````

While being an extremely simple type, booleans will turn to be essential in your program Python because they represent the logic that control the program behavior. Their primarily use is to compare things with the logical operators `<`, `>`, `<=`, `>=`, and `==`:

`````python
a = 3.2
b = 7.5
c = 7.5

print(a > b)
print(a < b)
print(a >= b)
print(a <= b)
print(a == b)
print(a != c)

print(b == c)
print(b > c)
print(b >= c)
`````

Be careful not to confuse the comparison operator `==` with the assignement operator `=`!

A logical reasonning can be implemented with the logical operators `and`, `or` and `not`: 

`````python
print( True or False )
print( False or False )
print( True or True )
print( True and False )
print( False and False )
print( True and True )
print(not True)
print(not False)
`````

For instance, to check if two conditions are met simultaneouly, one would write:

`````python
print( (3 > 2) and (-5 < -2) )
`````

Note that the parenthesis around each logical expression are not always mandatory, but makes the code clearer to read. Of course, logical operations can also be implemented with strings:

`````python
"ethanol" == "methanol"
`````

## Good practices

Excellent, you should now be able to define tons of variables within your future programs! 👍
Speaking of that, to be able to maintain clear and readable codes, it is important to follow a few "good practices" rules:
* You should always use variable names with a **clear signification**! So far in the lecture, variables were named `a`, `b`, `c`... you should never do that into your program, but use explicit names instead. For instance, store the string `"ethanol"` into a variable called `my_molecule_name`.
* **Do not use weird characters** (accent for instance) to name your variables. Stick to letters, numbers, and underscore character "_" (note however that you cannot start a variable name with a number).
* Try to use the **same naming convention everywhere**, for instance words separated by underscore, just like in `my_molecule_name`.

In the same vein, it is important that when you display something using `print()`, we know what it corresponds to. So far in the lecture, we used `print(my_variable)` to see the value of the variable `my_variable`: it is ok when dealing with a few lines of code, but rapidly gets incomprehensible in long program. Hopefully, there is a nice way to print things using so-called **f-strings**, which are just some kind of augmented strings. They are defined almost the same way as strings, only adding a `f` before the quotes:

`````python
my_fstring = f"My molecule is ethanol."
`````

`````python
print(my_fstring)
`````

As you can see it works exactly the same as string. Now, the cool things start when we introduce placeholders `{}`, where a variable will be replaced by its value:

`````python
my_molecule = "ethanol"
print(  f"My molecule is {my_molecule}" )
`````

Using f-strings is thus a quick way to get explicit print in your program! Note that you can put pretty much anything in the f-strings, as long as it is a valid Python expression:

`````python
my_molecule = "ethanol"
print(  f"My molecule is {my_molecule.upper()}" )
`````

In case you need to display numbers, it might also be useful to use the **modifiers** implemented in f-strings to control the way numbers are represented (decimal `f` or scientific notation `E`) and the number of digits you want to show (`.2` = 2 digits after the dot):

`````python
avogadro_constant = 6.02214076e23
print(f"The Avogadro constant is equal to { avogadro_constant} mol-1, which can also be written {avogadro_constant:.2f} mol-1 or {avogadro_constant:.2E} mol-1")
`````

As a very last thing before going to the next section, it is always a good idea to "annotate" your program to be sure that you are able to read it again later, or so that someone else can understand it. Also, if you give pieces of your code to an AI in order to solve some errors for instance, these annotations could help it. 
These annotations are better called **comments**, and they are simply introduced in the Python code starting the line with the `#` character: everything that follows is a comment, and therefore is ignored by the Python interpreter when the program is executed:

`````python
# Here we define the molecule names
molecule_name_1 = "ethanol"
molecule_name_2 = "methanol"

# Now we check if the molecule names are the same, and display the result to the screen
print(f"Are {molecule_name_1} and {molecule_name_2} the same molecule? {molecule_name_1 == molecule_name_2}")
`````
