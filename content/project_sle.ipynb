# Solid-Liquid Equilibrium phase Diagrams

In many areas of chemistry, materials science, and formulation science, understanding the solid–liquid equilibrium (SLE) of mixtures is essential. Solid–liquid phase diagrams describe how the melting behavior of a mixture depends on composition and temperature.

For a binary mixture of components A and B, the phase diagram typically shows:

- Liquidus curves, which represent the temperatures at which the first solid appears during cooling.
- Solid phases, where one or both components crystallize.
- A possible eutectic point, where the mixture melts at the lowest temperature and both solids coexist with the liquid phase.

The thermodynamic condition for equilibrium between a pure solid phase and a liquid mixture is based on the equality of chemical potentials, or Gibbs energy minimization. Several models can be used to calculate this equality between chemical potentials, including the simple ideal solution model and more advanced Gibbs energy excess models such as the Margule's model. In this project, you will build a program that allows the user to visualize the SLE of binary mixtures (i.e. a mix of two molecules).
  
You are free to build this project as you want, but below are a few tips that might be useful:

- Start with simle model (i.e. ideal solution), and then try to include more advanced models.
