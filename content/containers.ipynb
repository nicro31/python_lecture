# Containers

In the previous lecture section, you have learned about 4 Python types: float, int, string and bool. These are called Python primitive variables because they are the cornerstone of any program, but they are a bit limited to implement complex program structure. Non primitive variables are more advanced type of variables, that unlock the true power of Python and make possible to easily implement complex logical structure, in particular using so-called **loops**, but we will get to that later 😉. For now, let's discover these non primitive data types, that we could also call containers in the sense that they can store multiple values or omlex structures.

## List

The **list** type is the most common and useful kind of container. To continue the analogy with the computer memory as a chest of drawers, the list would be a deep drawer in which it is possible to store multiple values, as depicted below.

:::{figure} images/list.png
:align: center
:width: 100%

A list can be seen as a deep drawer in the computer memory, able of storing multiple values
:::

As you can notice, as a variable the list still has a **name**, `my_list` in that case, but it may hold multiple values, possibly of different types. The list depicted on the figure above is declared with the following line of code: 

```python
my_list = [-5.2, "hello", 8, False]
print(f"This is the type of the variable my_list: {type(my_list)}")
print(f"These are the values stored in the variable my_list: {my_list}")
```

The list is thus declared with brackets `[]` in which you write the values it holds separated by commas. If nothing is put inside the brackets, you still get a list but it's empty! 

```python
my_empty_list = []
print(f"This is the type of the variable my_list: {type(my_empty_list)}")
print(f"These are the values stored in the variable my_list: {my_empty_list}")
```

Then instead of declaring the list directly with the value it holds, you can first declare an empty list and progressively fill it... this can be extremely useful if at the beginning of the program you don't know yet what the list should contain! To add a value to the list, use the `append()` function as shown below:

```python
my_empty_list.append(-5.2)
print(f"These are the values stored in the variable my_list: {my_empty_list}")

my_empty_list.append("hello")
print(f"These are the values stored in the variable my_list: {my_empty_list}")

my_empty_list.append(8)
print(f"These are the values stored in the variable my_list: {my_empty_list}")

my_empty_list.append(False)
print(f"These are the values stored in the variable my_list: {my_empty_list}")
```

The drawer is progressively filled with values, starting from the "drawer front". Have you noticed how the `append()` function is called differently than the `print()` function for instance?

The `print()` function is a "regular" Python function, called with its name and some **arguments** in between the parentheses: the argument is the input of the function, the text we would like to print in this case. The `append()` function is what we called a **method**: this is still a function, but designed to work specifically with some kind of variable, the list in that case (indeed it would make no sense to try to append something to a float, that holds a single value). Therefore, it is called **from** the list by indicating its name, `my_empty_list`, followed by a dot `.`, and the method name, `append()`. In the parenthesis then we still find the input of the function (method), which is here the value you want to append to the list.
