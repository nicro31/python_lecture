# Classes

Last but not least, we must talk about **Classes** in this lecture, as you will manipulate many of these using Python, even if you don't really see it. The concept of **Class** is directly related to the concept of **Object-Oriented Programming (OOP)**, which is a programming paradigm at the core of many programming languages (Python, Java, C++...). In OOP, programs will manipulate so-called **objects**, which are some king of "super" variables that contain **both variables and functions**, respectively called **attributes and methods** in the case of objects. Actually, in Python everything is an object. For instance, `strings` are objects that make easier the text manipulation, and `list` are objects that make easier the manipulation of collection of variables.

Below is a minimalist example showing how a class is defined, using the `class` keyword:

`````python
class Molecule:

    # This special method defines the class attributes: here there is one attribute, the name
    def __init__(self, name):
        self.name = name

    # Define another method that print a formatted message giving the molecule name
    def print_molecule_name(self):
        print(f"This molecule name is {self.name}")
`````

Here are the main points to remember about class definition:

- A class is defined with the `class` keyword, followed by the class name you choose. Like for variables, keep a consistent way to write the class name throughout your program: in general, first letter is in upper case.
- Then, several methods can be defined using the `def` keyword, just like for functions. The methods always take, at least, `self` as an input argument. `self` refers to the current object, and is a variable that is therefore used to access the object attributes. But note that when you will use the class methods, you **do not need** to specify `self` as an argument as it is always implicit.
- The `__init__` method is a special method that must always be present, it will be called each time you create an object based on this class. Its purpose is to define the class atributes, i.e. what variables the class contain. Here the Molecule class contains one attribute, the Molecule name. It is stored in the self.name attribute. When a Molecule object is created, one must give `name` as an input of the `__init__` function, and this will be stored in the object.
- When you want to access a class attribute within the class definition, do no forget to write `self.` followed be the attribute, otherwise the interpreter will not understand that you refer to the class attribute!

Now, here is how one create an object based on the Molecule class:

`````python
molecule_1 = Molecule("methanol")
`````

As you can notice, no need to explicitely call the `__init__` method, just indicate the class name followed by the `__init__` input arguments if needed. The interesting point about classes is that once a class is defined, you can create as many objects as you like based on this class. Therefore, one can create a second molecule:

`````python
molecule_2 = Molecule("ethanol")
`````

And of course, every Molecule have the same methods, as they were defined based on the same template. The object methods are called by specifying the object name, followed by a dot `.`, and then the method name:

`````python
molecule_1.print_molecule_name()
molecule_2.print_molecule_name()
`````

The object attributes can be accessed the same way, with the object name followed by a dot `.`, and then the object attribute:

`````python
print(f"The first molecule name is {molecule_1.name}.")
print(f"The second molecule name is {molecule_2.name}.")
`````

class Molecule:

    # Note here that we will add an extra attribute to store the chemical element mass
    def __init__(self, smiles, name=""):
        self.smiles = smiles
        self.name = name
        self.mass_of_elements = {'H': 1.00794, 'C': 12.0107, 'N': 14.0067, 'O': 15.9994}

    def smiles_to_formula(self):

        # Get list of chemical elements only from smiles, i.e. remove SMILES special characters
        symbols = ''.join(char for char in self.smiles if char.isalpha())
    
        # Count number of atoms
        nb_atoms = len(symbols)
    
        # Count number of double or triple bonds to add
        nb_multiple = self.smiles.count("=") + 2 * self.smiles.count("#")
    
        # Deduce total number of bonds
        nb_bonds = nb_atoms - 1 + nb_multiple
    
        # Count theoretical valence from chemical elements in smiles
        valence = self.smiles.count("C") * 4 + self.smiles.count("O") * 2 + self.smiles.count("N") * 3
    
        # Finally infer the number of implicit hydrogen
        nb_H = valence - 2 * nb_bonds
    
        # Get the chemical formula as a string
        formula = ""
        if self.smiles.count("C") > 0:
            formula += f"C{self.smiles.count("C")}"
        if nb_H > 0:
            formula += f"H{nb_H}"
        if self.smiles.count("N") > 0:
            formula += f"N{self.smiles.count("N")}"
        if self.smiles.count("O") > 0:
            formula += f"O{self.smiles.count("O")}"
    
        # Return the chemical formula
        return formula
        

    def smiles_to_molar_mass(self):
    
        molar_mass = 0.
    
        # Note here how we will call a previously defined function to help us define the current function
        formula = self.smiles_to_formula()
    
        # Split the formula into pair of characters, representing chemical elements and number of occurences 
        result = []
        number_of_atoms = ""
        for char in formula:
            if char.isalpha():
                if number_of_atoms:
                    result.append((element, int(number_of_atoms)))
                    number_of_atoms= ""
                element = char
            else:
                number_of_atoms += char
    
        result.append((element, int(number_of_atoms)))
        
        for element, occurence in result:
            molar_mass += occurence * self.mass_of_elements[element]
    
        return molar_mass
        

    def is_alcohol(self):

        is_alcohol = False
    
        for i,char in enumerate(self.smiles):
            if char == "O":
                if i >= 1:
                    if self.smiles[i-1] == "C":
                        if i == len(self.smiles) - 1:
                            is_alcohol = True
                        if i < len(self.smiles) - 1:
                            if self.smiles[i] == ")":
                                is_alcohol = True
                        
                    if self.smiles[i-1] == "(":
                        if i >= 2:
                            if self.smiles[i-2] == "C":
                                if i == len(self.smiles) - 1:
                                    is_alcohol = True
                                if i < len(self.smiles) - 1:
                                    if self.smiles[i] == ")":
                                        is_alcohol = True
    
        return is_alcohol

    # Define a method that print summary of the molecule
    def print_molecule_summary(self):
        print(f"{10*"="}")
        print(f"This molecule name is {self.name}")
        print(f"SMILES: {self.smiles}")
        print(f"Chemical formula: {self.smiles_to_formula()}")
        print(f"Molar mass: {self.smiles_to_molar_mass():.5f} g/mol")
        print(f"Is alcohol: {self.is_alcohol()}")
        print(f"{10*"="}")
        print("\n")
