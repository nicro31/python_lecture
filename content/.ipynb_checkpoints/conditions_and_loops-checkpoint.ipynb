# Conditions and Loops

Variables are the cornerstone of any program: basically, the first question you should ask yourself when building a program is what variables, of which kind, do I need to solve my problem. But to be able to actually use these variables for complex tasks and to perform some operations, we also need to be able to structure our programs by implementing logics and automatization. This is the topic of this section.

## If statement

The first thing you would like to do in a program is to be able to implement some logical reasonning. That is to say, the program will so something in a scenario and something else in another. To do so, you can use the **if statement**. The `if` keyword is used to verify if a logical statement (a condition if you like) is true, and if it is the case it will perform the operation you want. Here is an example below:

```python
# Define two integer variables
my_int1 = 6
my_int2 = 3

# Do something only if my_int1 > my_int2
if my_int1 > my_int2:
    print(f"{my_int1} > {my_int2}")
```

Note the particular syntax of the if statement. It is followed by a semicolon `:`, and all instructions that are to be executed if the statement is true **must** be indented (there is some spaces between the beginning of the line and the `print`. You can choose any kind of indentations, but **you have to stick with the same indentation all over your program**. For instance, I use to put a tabulation as the indentation. Anything that follows the if and that are not indented are not part of the if statement! See for example what happens with the code below:

```python
# Define two integer variables
my_int1 = 3
my_int2 = 6

# Do something only if my_int1 > my_int2
if my_int1 > my_int2:
    print(f"{my_int1} > {my_int2}")
print(f"This line is printed anyway because not part of the if statement!")
```

In many cases, you would like to do something if the statement is true, but you also would like to do something in the case that the statement is false! To this end, you can use the `else` statement:

```python
# Define two integer variables
my_int1 = 3
my_int2 = 6

# Do something only if my_int1 > my_int2
if my_int1 > my_int2:
    print(f"{my_int1} > {my_int2}")
else:
    print(f"{my_int1} <= {my_int2}")
```

As you can see, the syntax stays the same: we need a semicolon `:` after the else statement, and everything that needs to be executed if the condition is false must be indented. Finally, one may also need to implement much complicated logics, when something should be done if a first condition is false, but only if a second consiiton is true. To this end, we can use the `elif` (stands for else if) statement:

```python
# Define two integer variables
my_int1 = 3
my_int2 = 3

# Do something only if my_int1 > my_int2, or in the case my_int1 <= my_int2, then do something only if my_int1 == my_int2. 
if my_int1 > my_int2:
    print(f"{my_int1} > {my_int2}")
elif my_int1 == my_int2:
    print(f"{my_int1} = {my_int2}")
else:
    print(f"{my_int1} < {my_int2}")
```

## For loops

Loops are used to repeat the same task many times. Basically, if you feel like you would need to do many copy/paste in your program to solve a problem, you probably need a loop! In particular, the `for` loop is used when you know beforehand how many times you need to repeat the operation. It is also used in conjunction with the `range()` function, that returns a list of consecutive numbers:

```python
for i in range(5):
    print(f"i = {i}")
```

Note that `i` is just a variable that "counts" the number of iteration. We could put any name for this variable, and it's a variable that exists only within the loop.
A very interesting aspect of the for loops is that you can iterate over the elements of a list directly:

```python
molecule_names = ["ethanol", "methanol", "water"]
for name in molecule_names:
    print(f"The name of the molecule is: {name}")
```

Using the `enumerate()` function, you get simultaneously the values in the list and their index:

```python
molecule_names = ["ethanol", "methanol", "water"]
for i,name in enumerate(molecule_names):
    print(f"The name of the molecule at index {i} is: {name}")
```

Note the particular syntax, with two variables separated by a comme: `i,name in...`. This is called unpacking. Since `enumerate` gives both values and index at the same time, i.e. two numbers, we need to unpack then in the two different variables `i` and `name`.

Since dictionnaries and lists are so close, you could imagine that you can also iterate over the elements of a dict. The most convenient syntax for that uses the `items()` method (function) of the dict:

```python
molecule_molecular_weight = {"ethanol": 46.068, "methanol": 32.04, "water": 18.01528}
for molecule_name,molecule_weight, in molecule_molecular_weight.items():
    print(f"The molecular weight of {molecule_name} is: {molecule_weight} g/mol")
```

## While loops

The `while` loop is a way to implement a loop when you don't know a priori the number of iterations you must perform. In a sense it combines the functionnalities of the `if` and the `for` statements, by doing the same task while a given condition is true. Here is an example:

```python
counter = 0
while counter < 5:
    print(f"counter = {counter}")
    counter += 1
```

Note that in this example, if you forget to increment the counter at each iteration, its value will always be 0 and the condition of the `while` will never be met, resulting in an infinite loop! This is a very common mistake that makes program crashes.
