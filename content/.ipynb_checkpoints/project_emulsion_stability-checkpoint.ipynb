# Simulation of Emulsion Stability

Emulsions are dispersions of droplets of one liquid phase in another immiscible liquid. They are widely used in many fields such as food science, cosmetics, pharmaceuticals, and materials science.

Over time, emulsions may become unstable due to several mechanisms as illustrated below. Droplets can migrate under gravity, leading to creaming (droplets rising) or sedimentation (droplets sinking). In addition, droplets undergo Brownian motion, can form aggregates, and sometimes coalesce to form larger droplets. Droplet sizes also evolve due to the so called Ostwald ripening. All these processes strongly influence the stability of the emulsion.

In this project, you will develop a numerical simulation to model the dynamics of droplets in an emulsion and visualize how phase separation emerges over time. To simplify the problem, we can deal with a 2D system (like in the figure below). The idea is to have a program where the parameters governing the emulsion stability can be modified, and one can observe the results with a graphical representation of the system.

You are free to build this project as you want, but below are a few tips that might be useful:
- The project basically involves two distinct functionnalities: simulating the emulsion stability and rendering a graphical visualisation.
- Your visual representation might be inspired by the figure below... but should evolve over time!
- To implement the different phenomena describing the droplets evolution over time, you need to do a bit of research to find out some simple equations that might describe them.
- Handle one phenomenon at a time, and check that it works fine alone before trying to combine several of them. I suggest you to start with creaming/sedimentation, Brownian motion, and then try to add the others if you have time.
