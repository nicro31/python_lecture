# Introduction and Instructions

Now that you know the basics of Python, it is time to create your first complete project! In the next lecture sections, you will find some project proposals. As you will see, they are meant to be vague. Indeed, since for this project you are free to use any tool, including AI, it means that the main difficulty will not be about the code writing itself, but more about how to think, manage and build a coding project.

## Deliverables

At the end of the project, you will deliver both your python program (basically one folder containing all your python files) and a small report. The report should contain the following:
- A first section dedicated to the presentation of the problem at hand, and in particular to the physical, chemical or mathematical knowledge required to solve the problem. For instance, if your program solve some equations at some points, they should be presented in this section.
- A second section describing the program by itself. You need to indicate which Python libraries you have used (so that anyone can reuse your program) and for what purpose. Also include a section dedicated to explain your program architecture. For instance, have you written several modules and for what purposes? What are the main classes used in the program? The main functions? I think that a graphical representation of the program architecture might be more revealing than a long text, but you are free to present this part as you like. Also explain (if applicable) how you have used AI to help yourself for this project; you can even include the prompts you used.
- If this apply to your program, include a section that will serve as a user manual. For instance, if your program include some buttons or menus to adjust variables, this should be explain so that anyone can use your program.
- Finally, show some results (screenshots) for instance demonstrating how cool your program is! In particular, discuss if the results you obtain seem reliable, in relation to the problem you are studying (e.g. if your program deals with some physicochemistry, show that the results are consistent).

## Advice

Here are a few tips to successfully carry out your project:
- Before writing any code lines, think globally about your project. How it should be organize, how you could share the work among your group.
- Start small, and adjust functionnalities step by step. Otherwise, you might end up with nothing working at the end.
- Be careful when using AI: it is a really powerful tool, but the risk is that you do many prompts in succession and at some points the program bugs and you have no idea why, because you have totally lost control over the program architecture.
- Save your project very regularly, and keep backups of your program all along its development!
- Start writing the report (a draft at least) alongside the program development, when you still keep in mind how the program precisely works.
- Do not hesitate to put many comments in your code so that from one session to the other your don't forget what a piece of code is doing.
