# Surfactant Selection

In emulsion formulation, the Hydrophilic-Lipophilic Balance (HLB) is a key parameter for selecting surfactants that stabilize oil-in-water (O/W) or water-in-oil (W/O) emulsions. Each oil has a required HLB (RHLB), and each surfactant has its own HLB, which can be estimated using Davies’ group contribution method for instance. Matching the surfactant mixture HLB to the oil blend’s RHLB optimizes interfacial tension, leading to a stable emulsion with proper droplet size and uniform distribution.
Formulators often need to:
- Select oils and surfactants from a large database.
- Calculate the RHLB of an oil blend.
- Determine surfactant blends that match the RHLB, optionally limiting the number of surfactants used.

In this project, you will develop a program that allows a user to manage a database of oils and surfactants. From these, the user will be able to create a formulation by selecting one or more oils to create a blend. The RHLB of the blend will be calculated, and based on this the program will propose a surfactant mixture (optionnaly with a specified number of components) whose HLB matches the formulation RHLB.

You are free to build this project as you want, but below are a few tips that might be useful:
- The core of the project is the database management and the optimization of the surfactant mix. Yet, it would be nice to have a user friendly graphical interface so that a formulator can use your program on a daily basis.
- To implement the project, you need to do a bit of research to find out how RHLB and HLB of blend can be calculated.
- It would be nice that you provide an initial database of oils and surfactants in your project, and then that users can extend this database.
