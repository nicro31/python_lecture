# Functions

We have seen previously that loops are a powerful syntax to automatize some tasks within a program. Another related concept is the **functions**, that will save you from having thousands of copy/paste in a complex program. Yet, they are quite different in the sense that functions will allow you to repeat a given task anywhere you need to in the program.

They are two steps when using functions in a program, the **function definition** and the **function call**. To start with a simple example, let's think about a python function just as a mathematical function. More precisely, we will construct a function that calculates the square of a number, i.e. *f(x) = x\*x*.

- The **function definition** is, as the name says, a definition of what a function should do. It has to be defined once and for all. The syntax is as follows, and uses the `def` keyword (stands for definition):

`````python
def square(x):
    square_of_x = x*x
    return square_of_x
`````

As you can see, all the content of the function should be indented, just like for the loops. The keyword `def` is followed by the name of the function, `square` in this example. You can name the functions as you like, just be sure to respect the general rules that holds for variables naming. The function name is **always** followed by parenthesis `()`, in which we will put the function **arguments**. An argument is a variable (of any type) that will be the input of the function, i.e. that we will use to "do thing" in the function. Note that there may be no argument, one argument, or multiple arguments. Finally, the line end with the semicolon `:`, just like for loops.



Then in the function body, you can perform any operations you woulk like, even with complex logical structures, loops and so-on. In this example, we just define a basic mathematical operation involving the `x` variable. At the end of the function definition, you will find the `return` keyword. This keyword indicates that the function should stop and returns the variable (or variables) following `return`.

- The **function call** is when you actually ask the program to execute the function. To do so, you just need to write the function name, with the input variables in between parenthesis. Of course, the function needs to be defined before it can be called! To calculate the square of `3`, what would thus do:

`````python
# Define a variable that holds a number
my_number = 3

# Calculate the square of my_number and store that in a variable called my_number_squared
my_number_squared = square(my_number)

# Print the formatted result
print(f"The square of {my_number} is {my_number_squared}")
`````

Note that your input variable in the function call (`my_number` here) **doesn't need to be named the same as the function argument in the function defition**. Actually, one could directly call the `square` function with a number as the input: 

`````python
# Directly calculate and print the square of 3
print(f"The square of 3 is {square(3)}")
`````

This is because the function definition acts as some kind of template for the program. When you call the `square` function with an input variable, let's say a variable named `my_number` (or directly a number), the value of the variable (or the number itself) is copied into a **local variable** called `x`. Then some operations are performed **locally** with this `x`, and the functions returns something, exits, and `x` is destroyed: it has only a temporary lifetime during the function call, outside **it does not exist** (or its behavior will be uncertain at least). The same holds for any variables defined within the function: they are local variables, and therefore should not be accessed out of it.

Take a look at the example below to check out this concept of local variables:

`````python
# Function definition, using local variables x and square of x
def square(x):
    square_of_x = x*x # Calculating the square of x, and store in a local variable sqaure_of_x
    x = 2*x # Modifying the local variable x
    print(f"This is a message from the square function, local x = {x}")
    return square_of_x

# Define variables x and square_of_x out of the scope of the function sqaure
x = 3
square_of_x = 0

# Call the function
square_of_x_after_function = square(x)

# Print all variables
print(f"global x = {x}")
print(f"square_of_x = {square_of_x}")
print(f"square_of_x_after_function = {square_of_x_after_function}")
`````

As you see, the *global* `x` is unchanged while the *local* x in the function definition is modified. The global `square_of_x` variable is also left unchanged, while the *local* `square_of_x` contains the actual square of x!

As stated previously, a function definition can involve multiple arguments. In that case, during the function call, every input variable is mapped to the function argument in their order of appearance. For instance, let's define a function that just prints its two arguments:

`````python
# Define the function
def print_arg(x1, x2):
    print(f"The function arguments are {x1} and {x2}")

# Define two variables
my_var1 = 3
my_var2 = "hello!"

# Call the function
print_arg(my_var1, my_var2)

# Call the function with "inverted" input parameters
print_arg(my_var2, my_var1)
`````

Note that the `print_arg` function doesn't have a `return` keyword: this is fine, in that case the function does something but doesn't output a value at the end. There is also the possibility to use **default arguments** as function arguments. In a function call, arguments must always be input, while default arguments may not be; in that case, their value will be a predefined value defined in the function definition, as shown below: 

`````python
# Define the function with a default argument
def print_arg(x1, x2, x3 = "this is the default argument"):
    print(f"The function arguments are {x1} and {x2} and {x3}")

# Define two variables
my_var1 = 3
my_var2 = "hello!"
my_var3 = True

# Call the function with an input for the default argument
print_arg(my_var1, my_var2, my_var3)

# Call the function without an input for the default argument
print_arg(my_var1, my_var2)
`````

````{admonition} Exercice
:class: tip
:label: exo_function_1

Define a function that allows the user to calculate a x to the power of n (n being an integer). By default, the function should calculate the power of two if the user is to lazy to input the power exponent. Of course, test the function on several examples.

````

````{dropdown} Solution
```{code-cell}

# Define the function
def power(x,n=2):
    result = 1
    for i in range(n):
        result *= x
    return result

# Test the function with all parameters
x = 4
n = 0
print(f"{x}^{n} = {power(x,n)}")

# Test the function with all parameters
x = 4
n = 1
print(f"{x}^{n} = {power(x,n)}")

# Test the function with all parameters
x = 4
n = 3
print(f"{x}^{n} = {power(x,n)}")

# Test the function with default parameters
x = 4
print(f"(Default parameter) {x}^{2} = {power(x)}")

```
````
