# Graphs with Matplotlib

`matplotlib` is the most popular library to plot graphs using Python. It is really easy to use and gives many possibilities to adjust the rendering. To start with this lecture section, install the `matplotlib` library using `conda install matplotlib` in your python environment. Then, to be able to actually plot thing, import the `pyplot` module of the library as follows:

```python
import matplotlib.pyplot as plt
```

## Basic plots

To illustrate the plotting capabilities of `matplotlib`, let's first prepare some random data to plot. In this example, we will plot the function $f(x) = x^2$. 

```python
# Generate the data
x = [i*0.1 for i in range(10)]
y = [xx**2 for xx in x]
```

With the data at hand, generating a basic plot only takes one line of code:

```python
# Plot the data
plt.plot(x,y)

# Show the plot (Optionnal in Jupyter Notebook as the plot is automatically shown, but needed if we want to plot several curves in the same code cell)
plt.show()
```

Easy, I told you! Yet you know that a plot without axis labels is totally meaningless in sciences, let's correct this:

```python
# Plot the data
plt.plot(x,y)

# Add axis label and a title
plt.title("Square Function")
plt.xlabel("x")
plt.ylabel("f(x)")

# Show the plot (Optionnal in Jupyter Notebook as the plot is automatically shown, but needed if we want to plot several curves in the same code cell)
plt.show()
```

Ah that is much better! Now, you might want to tune a bit the plot to make it prettier. `matplotlib` as many options to do so, but we definitely don’t have time to cover this in this course. Just like for pandas, if you want to do a certain modification, just search the Web or ask an AI, you will find the solution quite easily. To convince you that `matplotlib` can render quality graphs, take a look at the modified plot below: 

```python
# Adjust Figure size
plt.figure(figsize=(7, 5))

# Plot with markers and dotted line
plt.plot(
    x, y,
    linestyle="--",      # dotted line
    marker="o",         # marker
    markersize=6,
    linewidth=1.0,
    color=(0.0,0.0,0.0) # Black color
)

# LaTeX style axis label
plt.xlabel(r"$x$", fontsize=18)
plt.ylabel(r"$f(x) = x^2$", fontsize=18)

# Title
plt.title(r"$Square \ Function$", fontsize=20)

# Add a grid
plt.grid(True, linestyle="--", linewidth=0.6, alpha=0.7)

# Axis ticks adjustment
plt.minorticks_on()
plt.tick_params(
    axis='both',           
    direction='in',        
    top=True,              
    right=True,            
    length=6,              
    width=1,               
    labelsize=12
)
plt.tick_params(
    axis='both',
    which='minor',
    direction='in',
    top=True,
    right=True,
    length=3,       
    width=0.8
)

# Aspect ratio of the plot
plt.axis("square")


# Axis limit
plt.xlim([0.0,1.0])
plt.ylim([0.0,1.0])


# Show the final plot
plt.show()
```

## Using matplotlib with pandas

As `pandas` is the perfect tool to manage numerical data, it would be nice to be able to plot data from a `DataFrame`. This is actually quite straightforward. To illustrate how it works, load the data prepared in the previous section that you should have saved in an excel file:

```python
import pandas as pd
df = pd.read_excel("filtered_data.xlsx")
df.head(10)
```

Now, here is how to plot some specific columns of the `DataFrame`:

```python
# Plot the data
plt.plot(df["time"],df["output"])

# Add axis label and a title
plt.title("Filtered Data")
plt.xlabel("time")
plt.ylabel("output")

# Show the plot
plt.show()
```

`matplotlib` also offers the possibility to have several graphs on the same Figure, which can be useful if you need to plot several data at the same time:

```python
# Figure size
plt.figure(figsize=(12,5))

# First subplot (1 line, 2 columns, position 1)
plt.subplot(1, 2, 1)
plt.plot(df["time"], df["output"], linestyle=':', marker='o', color='darkorange')
plt.xlabel(r"$time$")
plt.ylabel(r"$output$")
plt.title("First Data")
plt.grid(True)

# Second subplot (1 line, 2 columns, position 2)
plt.subplot(1, 2, 2)
plt.plot(df["time"], df["squared output"], linestyle='--', marker='s', color='steelblue')
plt.xlabel(r"$time$")
plt.ylabel(r"$squared \ output$")
plt.title("Second Data")
plt.grid(True)

# Show plot 
plt.show()

```

Or you can choose to plot the two curves on the same graph. In that case I advise you to plot the graph using a slightly diffrent syntax that uses the axes object of the plot, as shown below. Do not forget to add a legend to know which curve corresponds to which data!

```python
# Création de la figure
fig, ax = plt.subplots(figsize=(7, 5))

#First plot
ax.plot(
    df["time"], df["output"],
    linestyle=":",      
    marker="o",         
    markersize=1,
    linewidth=1.,
    color=(0.0,0.0,0.0)
)

# Seond plot on the same axis
ax.plot(
    df["time"], df["squared output"],
    linestyle="--",     
    marker="+",         
    markersize=1,
    linewidth=1.,
    color=(0.75, 0.75, 0.75)
)

# LaTeX style axis label
ax.set_xlabel(r"$time$", fontsize=18)
ax.set_ylabel(r"$signal$", fontsize=18)

# Title
ax.set_title(r"$Filtered \ Data$", fontsize=20)

# Add a grid
ax.grid(True, linestyle="--", linewidth=0.6, alpha=0.7)

# Axis ticks adjustment
ax.minorticks_on()
ax.tick_params(
    axis='both',           
    direction='in',        
    top=True,              
    right=True,            
    length=6,              
    width=1,               
    labelsize=12
)
ax.tick_params(
    axis='both',
    which='minor',
    direction='in',
    top=True,
    right=True,
    length=3,       
    width=0.8
)
# y limit of the axis
ax.set_ylim([-1.,1.])

# Add legend
ax.legend(["output", "squared output"])

# Show the final plot
plt.show()
```

## 3D plots

We will not go into much details here, but just so that you know, `matplotlib` can also handle 3D plots, a capability that can be quite useful in many cases. For instance, let's say that you want to plot the function $f(x,y) = cos(x) * sin(x)$. To generate the data associated to this function, we will use a library called `numpy`, very popular, dedicated to the manipulation of multidimensionnal arrays or matrices. Install the library with `conda install numpy`. Then, to represent the data as 3D point cloud, do as follows:

```python
# Create the data
import numpy as np
x = np.arange(-5, 5, 0.25) # Create an array of number between -5 and 5, with a step of 0.25
y = np.arange(-5, 5, 0.25)
x, y = np.meshgrid(x, y) # From the array X and Y previously defined, generate new arrays that map the entire space delimited by X and Y 
z = np.cos(x) * np.sin(y)

# Create a figure
fig = plt.figure()

# Create a plot with a 3D view
ax = fig.add_subplot(projection='3d')

# Plot the data as a point cloud
ax.scatter(x, y, z, marker="o")

# Label axis
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("z")

# Show the plot
plt.show()
```

You can instead represent the data as a surface:

```python
# Give access to colormap
from matplotlib import cm

# Create a figure
fig = plt.figure()

# Create a plot with a 3D view
ax = fig.add_subplot(projection='3d')

# Plot the data as a surface
surf = ax.plot_surface(x, y, z, cmap=cm.coolwarm,linewidth=0, antialiased=False)

# Label axis
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("z")

# Show the plot
plt.show()
```

Or as a 2D colored map, with some contour lines for instance:

```python
# Create a figure
fig = plt.figure()

# Create a plot with a 2D view
ax = fig.add_subplot()

# Plot the data as a surface
cs = ax.contourf(x, y, z, 20, cmap=cm.coolwarm)

# Label axis
ax.set_xlabel("x")
ax.set_ylabel("y")

# Add a color bar
cbar = fig.colorbar(cs)

# Show the plot
plt.show()
```

## Miscellaneous

As a chemist, you may want at some point to draw high quality figures that combine both numerical data and molecular structure. This can be easily done y combining the power of `rdkit` and `matplotlib`, as illustrated below:

```python
# Import rdkit modules
from rdkit import Chem
from rdkit.Chem import Draw

# Import modules to add image to plot
from matplotlib.offsetbox import OffsetImage, AnnotationBbox

# Create a molecule from a SMILES code
mol = Chem.MolFromSmiles("OC(=O)CN(CCN(CC(O)=O)CC(O)=O)CC(O)=O")

# Generate an image of the molecular structure
img = Draw.MolToImage(mol, size=(600,600), imageType='RGBA')

# This small code section modifies the molecule image
# so that it gets a slightly transparent background instead of a white background
# if we don't, we end up with a white rectangle on the final plot
img = img.convert("RGBA")
datas = img.getdata()
newData = []
for item in datas:
    # item = (R,G,B,A)
    if item[0] > 254 and item[1] > 254 and item[2] > 254:  # presque blanc
        newData.append((255,255,255,100))  # transparent
    else:
        newData.append(item)
img.putdata(newData)

# Plot your numerical data
fig = plt.figure()
ax = fig.add_subplot()
cs = ax.contourf(x, y, z, 20, cmap=cm.coolwarm)
ax.set_xlabel("x")
ax.set_ylabel("y")
cbar = fig.colorbar(cs)

# Create an OffsetImage
im = OffsetImage(img, zoom=0.25)  # zoom adjust the image size

# Create an AnnotationBbox to position on the image
ab = AnnotationBbox(im, (2, 1.75), frameon=False)  # position (x,y)
ax.add_artist(ab)

plt.show()
```

Finally, note that you can save your figures in order to be able to use them elsewhere, e.g. to write a report. The saving is straightforward with the `savefig` method:

```python
fig.savefig("figure.pdf", dpi=300, bbox_inches='tight', transparent=True) # Here we can use other image format, such as png, svg...
```
