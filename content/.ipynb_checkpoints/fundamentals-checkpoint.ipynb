## Introduction

You are now an expert in Python and you can write programs to solve complex tasks! 😎
But as you have seen throught the exercices you had to solve so far, even a relatively simple task can require quite a large amount of code. For instance, let's say that you need to calculate some cosines in your program. Well, you would probably define a function to be able to do so easily in your code wherever you need to. This can be done using the Taylor's expansion of a cosine: 

$$
\cos(x) = \sum_{n=0}^{\infty} \frac{(-1)^n x^{2n}}{(2n)!}
$$

Based on this developement, the cosine function can be defined as follows:

`````python
def cosine(x, nb_terms=20):
    """
    Calculate an approximation of cos(x) (in radians)
    using the Taylor's expansion
    
    x : angle in radians
    nb_terms : number of terms used in the expansion (the higher the better)
    """
    result = 0
    sign = 1
    power = 1  # x^(2n)
    factorial = 1  # (2n)!
    
    for n in range(nb_terms):
        if n > 0:
            # Update x^(2n)
            power *= x * x
            # Update (2n)!
            factorial *= (2*n - 1) * (2*n)
        
        result += sign * (power / factorial)
        sign *= -1
    
    return result
`````

Let's try our cosine function:

`````python
# Define pi
pi = 3.14159265359

# Check the function on particular angle
print(f"Cosine of 0 degree is {cosine(0.0 * pi / 180.0)}")
print(f"Cosine of 30 degree is {cosine(30.0 * pi / 180.0)}")
print(f"Cosine of 60 degree is {cosine(60.0 * pi / 180.0)}")
print(f"Cosine of 90 degree is {cosine(90.0 * pi / 180.0)}")
print(f"Cosine of 180 degree is {cosine(180.0 * pi / 180.0)}")
`````

Excellent, these are accurate values of the cosine! Yet this is quite a large piece of code just for cosine... Imagine if you have to define yourself all the basic mathematical functions, what a pain! In addition, there are for sure many people in the worlds that also need such functions in their code, therefore everyone is writing the same code over and over... 
Hopefully, the Python community is a friendly community, and many functions you may need were already written and shared by people. 😉

These bunches of functions are shared through what is called **libraries** in Python. A **library** is a complete Python project, distributed and installable, and dedicated to solve a specific set of tasks. In the following, you will for instance discover libraries to manipulate molecules, to handle data table, and to plot graph. There are **standard Python libraries**, shipped with the Python distribution you have downloaded at the beginning of this lecture, and **third-party libraries**, usually installed with the `conda install` or `pip install` commands. Once a library is made available in your Python environment, it is really straightforward to use. Let's see an example using the standard `math` library, dedicated to mathematical operations. 

To be able to use the library in your program, you need to **import** it, using the `import` keyword:

`````python
import math
`````

And that's it, from now on you can use all functions defined in this library!

`````python
# Calculate a cosine using the math library
cos_30 = math.cos( 30.0 * math.pi / 180.0 )
print(f"Cosine of 30 degree is {cos_30}")
`````

As you can see, the `math` library not only contains functions but also variable definitions, such as `math.pi`. Actually, a library can contain anything: variable definitions, functions, classes... It is possible to look at a library content using the `help` function:

`````python
# Check-out the content of the math library
help(math)
`````

Note also that you can import all functionnalities of a library so that you don't need to specify the library name everywhere in the code:

`````python
# This import everything from math, and remove the need to specify the library name in the code afterwards
from math import*

sin_30 = sin( 30.0 * pi / 180.0 )
print(f"Sine of 30 degree is {sin_30}")
`````

At this point you should be convinced by the usefulness of the libraries, it almost feel magical to be able to do complex calculations with just one line of code! But it has nothing magical, this is just the result of the work of people all around the world, that wrote pieces of code and share them. By the way, how do we write and share a library? Well, this is done through what are called **packages** and **modules** in Python, which are very convenient ways to organize large amount of code such as libraries. A **module** is just a Python file (with the `.py` extension), where functions, classes and so on are written. A **package** is a folder on the computer where several **modules**, generally used in conjunction to handle specific sub-tasks, are stored. Then, a **library** is just an ensemble of organized packages. This hierarchical organization of the code is really powerful as it facilitates the code maintenance and helps to navigate large codebases. It also makes possible to have several people working on the same project, each person being for instance responsible of one specific module.

## Write your own modules and packages

I really advise you to try to organize your code in modules and packages when working on large project. Actually, this is really easy! 
To create your first package, follow these steps:
- Create a folder called `my_package` next to the jupyter notebook you are using.
- In this folder, create two Python files, with extension `.py`: call one file `my_variable.py` and the other `my_function.py`.
- In `my_variable.py`, just define a variable `name`, that contains your first name.
- In `my_function.py`, just define a function that takes a string as input argument and print `My name is` followed by the input string.
- That's it, you have created your first package, `my_package`, containing two modules, `my_variable` and `my_function`!
Now, to use your package, it works just like for libraries:

`````python
# Code for finding the package, only to be used if the package is not located in the same directory as the jupyter notebook (in that case, indicates where the package is located relatively to the notebook)
import sys
import os
sys.path.append(os.path.abspath("./content"))
`````

`````python
# We first import the modules by specifying the package name followed by a dot, and then the module name 
import my_package.my_variable
import my_package.my_function

# Now the modules can be used: to access the module names and variables, you still need to specify the module they belong to
my_package.my_function.print_name(my_package.my_variable.name)
`````

To lighten the code writing, you can use again the shortcut to import everything from the modules:

`````python
from my_package.my_variable import*
from my_package.my_function import*
print_name(name)
`````
